```python
import pandas as pd
```

```python
#combine yellow taxi dfs
y1 = pd.read_parquet('yellow_tripdata_2025-01.parquet')
y2 = pd.read_parquet('yellow_tripdata_2025-02.parquet')
y3 = pd.read_parquet('yellow_tripdata_2025-03.parquet')
y4 = pd.read_parquet('yellow_tripdata_2025-04.parquet')
y5 = pd.read_parquet('yellow_tripdata_2025-05.parquet')

yellow_taxi = pd.concat([y1,y2,y3,y4,y5], axis=0)
yellow_taxi.to_parquet('yellow_tripdata_2025.parquet', index=False)
```

```python
#combine green taxi dfs
g1 = pd.read_parquet('green_tripdata_2025-01.parquet')
g2 = pd.read_parquet('green_tripdata_2025-02.parquet')
g3 = pd.read_parquet('green_tripdata_2025-03.parquet')
g4 = pd.read_parquet('green_tripdata_2025-04.parquet')
g5 = pd.read_parquet('green_tripdata_2025-05.parquet')

green_taxi = pd.concat([g1,g2,g3,g4,g5], axis=0)
green_taxi.to_parquet('green_tripdata_2025.parquet', index=False)
```

```python
#combine fhvhv dfs
f1 = pd.read_parquet('fhvhv_tripdata_2025-01.parquet').drop(
    ['bcf', 'access_a_ride_flag', 'wav_request_flag', 'wav_match_flag', 'dispatching_base_num', 'originating_base_num'],
    axis=1,
    errors='ignore'
)
f2 = pd.read_parquet('fhvhv_tripdata_2025-02.parquet').drop(
    ['bcf', 'access_a_ride_flag', 'wav_request_flag', 'wav_match_flag', 'dispatching_base_num', 'originating_base_num'],
    axis=1,
    errors='ignore'
)
f3 = pd.read_parquet('fhvhv_tripdata_2025-03.parquet').drop(
    ['bcf', 'access_a_ride_flag', 'wav_request_flag', 'wav_match_flag', 'dispatching_base_num', 'originating_base_num'],
    axis=1,
    errors='ignore'
)

fhvhv_trip = pd.concat([f1,f2,f3], axis=0)
fhvhv_trip.to_parquet('fhvhv_tripdata_2025.parquet', index=False)
```
